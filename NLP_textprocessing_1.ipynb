```python
import nltk
import string
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('punkt_tab')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
text = """
My name is Irine and I come from Perumbavoor, Ekm.
I completed my schooling from MGM HSS, Kuruppampady, Eranakulam.
My family consists of 4 members including my parents, brother and me.
I am currently pursuing my final year of B.Tech in Artificial Intelligence at Adi Shankara Institute, Kalady.
My achievements are, I have won several Medals and certificats in Karate and football.
During my college, I have completed a mini project and currently working on main project.
My Hobbies include reading books, learning languages and driving.
My interests are travelling, coding and experiencing new things.
My strenghts include adaptivity, leadership qualities and communication skills.My weaknesses are, I am a slow learner and overthinks a lot.
"""
print("ORIGINAL TEXT\n")
print(text)

l_text = text.lower()
print("\n1. LOWERCASE\n")
print(l_text)

sentences = sent_tokenize(l_text)
print("\n2. SENTENCE TOKENIZATION\n")
print(sentences)

words = word_tokenize(l_text)
print("\n3. WORD TOKENIZATION\n")
print(words)

words_no_punct = [word for word in words if word not in string.punctuation]
print("\n4. AFTER REMOVING PUNCTUATION\n")
print(words_no_punct)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words_no_punct if word not in stop_words]
print("\n5. AFTER REMOVING STOPWORDS\n")
print(filtered_words)

ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]
print("\n6. STEMMING\n")
print(stemmed_words)

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("\n7. LEMMATIZATION\n")
print(lemmatized_words)

final_text = " ".join(lemmatized_words)
print("\n8. FINAL PREPROCESSED TEXT\n")
print(final_text)
```
